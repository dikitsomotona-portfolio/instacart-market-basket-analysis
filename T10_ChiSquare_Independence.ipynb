**Research Question**

Is reorder behaviour independent of day of week?

H0 : Reorder behaviour is independent of day of week

Ha : Reorder behaviour depends on day of week

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from google.colab import drive

drive.mount('/content/drive', force_remount=True)

interim_path = '/content/drive/MyDrive/ADA2026_Instacart_Project/Data/Interim/'

orders = pd.read_csv(interim_path + 'orders_interim.csv')
products = pd.read_csv(interim_path + 'products_interim.csv')
order_products = pd.read_csv(interim_path + 'order_products_interim.csv')

print("Data loaded successfully ")
print(f"Orders: {orders.shape}")
print(f"Products: {products.shape}")
print(f"Order Products: {order_products.shape}")
```

```python
# Step 1: Construct Contingency Table

# Merge reordered information with orders dataset
merged = order_products.merge(
    orders[['order_id', 'order_dow']],
    on='order_id'
)

# Create contingency table
ct = pd.crosstab(merged['order_dow'], merged['reordered'])

ct
```

```python
chi2, p, dof, expected = stats.chi2_contingency(ct)

print("Chi-square statistic:", chi2)
print("P-value:", p)
print("Degrees of freedom:", dof)
```

```python
# Step 3: Examine Expected Frequencies

expected
```

**Interpretation**

The Chi-Square test results are used to determine whether reorder behaviour is independent of the day of the week. If the p-value is less than 0.05, the null hypothesis is rejected, indicating that there is a statistically significant association between day of the week and reorder behaviour. This means customers’ likelihood of reordering products changes depending on the specific day. If the p-value is greater than or equal to 0.05, we fail to reject the null hypothesis, meaning there is no sufficient evidence to conclude that reorder behaviour depends on the day of the week.

**Operational Implications**

If a significant relationship exists, Instacart can use this insight to improve demand forecasting by adjusting inventory levels based on different days of the week. This would help reduce stockouts and improve delivery efficiency during high-demand days. It can also support better workforce scheduling and targeted promotions to maximise sales. If no relationship is found, Instacart can maintain a consistent operational strategy across all days, focusing instead on other factors that influence reorder behaviour such as product type or customer preferences.

## **Task 1: Department vs Reorder Behaviour**

**Research Question:**
Is reorder behaviour independent of product department?

**Hypotheses:**
- H₀: Reorder behaviour is independent of product department

- Hₐ: Reorder behaviour is associated with product department

**Significance Level:** α = 0.05

**Degrees of Freedom:** df = (r - 1)(c - 1)

**Assumptions:**
1. Observations are independent
2. Categories are mutually exclusive
3. Each expected frequency is at least 5

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from google.colab import drive

drive.mount('/content/drive', force_remount=True)

interim_path = '/content/drive/MyDrive/ADA2026_Instacart_Project/Data/Interim/'

orders = pd.read_csv(interim_path + 'orders_interim.csv')
products = pd.read_csv(interim_path + 'products_interim.csv')
order_products = pd.read_csv(interim_path + 'order_products_interim.csv')

print("Data loaded successfully ")
print(f"Orders: {orders.shape}")
print(f"Products: {products.shape}")
print(f"Order Products: {order_products.shape}")
```

**STEP 1 : Construct Contingency Table**

```python
# Merge order_products with products to get department_id
merged = order_products.merge(products[['product_id', 'department_id']],
                               on='product_id', how='inner')

# Construct contingency table
ct = pd.crosstab(merged['department_id'], merged['reordered'])
ct.columns = ['First-Time', 'Reorder']

print("Contingency Table:")
print(ct)
print(f"\nShape: {ct.shape}")
```

**STEP 2 & 3 : Compute expected frequencies and Perform chi - square test**

```python
# Perform chi-square test
chi2, p, dof, expected = stats.chi2_contingency(ct)

# Display expected frequencies table
expected_df = pd.DataFrame(
    expected,
    index=ct.index,
    columns=ct.columns
)
print("Expected Frequencies:")
print(expected_df)

print(f"Chi-Square Statistic: {chi2:.4f}")
print(f"p-value: {p}")
print(f"Degrees of Freedom: {dof}")

print(f"\nDecision at α = 0.05:")
if p < 0.05:
    print("Reject H₀  — reorder behaviour is associated with department")
else:
    print("Fail to Reject H₀ — reorder behaviour is independent of department")

# Verify expected frequencies
print(f"\nMinimum Expected Frequency: {expected.min():.2f}")
print(f"All expected frequencies ≥ 5: {(expected >= 5).all()}")
```

**STEP 4 : Identify cells contributing most to χ2**

```python
# Compute contributions
contributions = (ct.values - expected)**2 / expected
contrib_df = pd.DataFrame(contributions,
                          index=ct.index,
                          columns=ct.columns)

# Top contributing cells
contrib_stack = contrib_df.stack().reset_index()
contrib_stack.columns = ['department_id', 'reorder_flag', 'contribution']
contrib_stack = contrib_stack.sort_values('contribution', ascending=False)

print("Top 5 cells contributing most to χ²:")
print(contrib_stack.head(5))
```


**Business Interpretation:**

Department 13 contributes most to the χ² statistic, indicating that its reorder behaviour deviates strongly from what would be expected under independence. Departments 16 and 4 also show notable deviations, suggesting differences in purchasing patterns across these departments. Department 11 shows higher-than-expected first-time purchases, implying more trial behaviour from customers.

From a business perspective, Instacart can use this to target promotions: improve repeat purchase strategies in low-reorder departments and strengthen customer retention in departments with unusual purchasing behaviour.

## **Task 2: Hour of Day vs Weekend Indicator**

**Research Question:**
Is the hour of day when orders are placed associated with whether the order occurs on a weekend or weekday?

**Hypotheses:**

- H₀: Order hour and weekend/weekday indicator are independent

- Hₐ: Order hour and weekend/weekday indicator are associated

**Significance Level:** α = 0.05

**Degrees of Freedom:** df = (r - 1)(c - 1) where r = number of hour categories and c = 2 (weekend vs weekday)

**Assumptions:**
1. Observations are independent
2. Categories are mutually exclusive
3. Each expected frequency is at least 5

**STEP 1 : Create weekend indicator variable**

```python
orders['weekend'] = orders['order_dow'].apply(
    lambda x: 1 if x in [0, 6] else 0
)
```

**STEP 2 : Cross - tabulate with order hour**

```python
ct2 = pd.crosstab(
    orders['order_hour_of_day'],
    orders['weekend']
)

ct2.columns = ['Weekday', 'Weekend']

print("Contingency Table: Hour of Day vs Weekend Indicator")
print(ct2)
```

**STEP 3 : Perform chi-square test**

```python
# Perform chi-square test
chi2_2, p_2, dof_2, expected_2 = stats.chi2_contingency(ct2)

print(f"Chi-Square Statistic: {chi2_2:.4f}")
print(f"p-value: {p_2}")
print(f"Degrees of Freedom: {dof_2}")
print(f"\nDecision at α = 0.05:")
if p_2 < 0.05:
    print("Reject H₀  — hour of day is associated with weekend indicator")
else:
    print("Fail to Reject H₀ — hour of day is independent of weekend indicator")

# Verify expected frequencies
print(f"\nMinimum Expected Frequency: {expected_2.min():.2f}")
print(f"All expected frequencies ≥ 5: {(expected_2 >= 5).all()}")

# Top contributing cells
contributions_2 = (np.array(ct2) - expected_2)**2 / expected_2
contrib_df2 = pd.DataFrame(contributions_2,
                            index=ct2.index,
                            columns=ct2.columns)
contrib_stack2 = contrib_df2.stack().reset_index()
contrib_stack2.columns = ['hour', 'day_type', 'contribution']
contrib_stack2 = contrib_stack2.sort_values('contribution', ascending=False)

print("\nTop 5 cells contributing most to χ²:")
print(contrib_stack2.head(5))
```


**Business Interpretation:**

The chi-square test indicates a significant association between hour of day and weekend indicator, suggesting that ordering patterns differ between weekdays and weekends. Early morning hours (e.g., 6–7 AM) show the highest χ² contributions, indicating lower-than-expected weekend activity during these hours compared to weekdays.

From a business perspective, Instacart should adjust staffing and delivery schedules to align with these time-based demand differences. Marketing campaigns could also target weekend shoppers during mid-morning to afternoon hours when order activity is highest

## **Task 3: Order Size Category vs Reorder Behaviour**

**Research Question:**
Is reorder behaviour associated with order size category?

**Hypotheses:**
- H₀: Reorder behaviour is independent of order size category
- Hₐ: Reorder behaviour is associated with order size category

**Significance Level:** α = 0.05

**Degrees of Freedom:** df = (r - 1)(c - 1)

**Order Size Categories:**
- Small: 1 — 5 products
- Medium: 6 — 15 products
- Large: 16+ products

**Assumptions:**
1. Observations are independent
2. Categories are mutually exclusive
3. Each expected frequency is at least 5

**STEP 1 : Categorise order size into small , medium , large**

```python
# Construct order size
order_size = order_products.groupby('order_id').size().reset_index(name='order_size')

# Merge with order_products
data = order_products.merge(order_size, on='order_id', how='inner')

# Categorise order size
def categorise_order(x):
    if x <= 5:
        return 'Small'
    elif x <= 15:
        return 'Medium'
    else:
        return 'Large'

data['order_size_cat'] = data['order_size'].apply(categorise_order)

print("Order Size Category Distribution:")
print(data['order_size_cat'].value_counts())
```

**STEP 2 & 3 :  Construct contingency table with reorder flag & Perform Chi-square test**

```python
# Construct contingency table
ct3 = pd.crosstab(data['order_size_cat'], data['reordered'])
ct3.columns = ['First-Time', 'Reorder']
ct3.index.name = 'Order Size Category'

print("Contingency Table:")
print(ct3)

# Perform chi-square test
chi2_3, p_3, dof_3, expected_3 = stats.chi2_contingency(ct3)

print(f"\nChi-Square Statistic: {chi2_3:.4f}")
print(f"p-value: {p_3}")
print(f"Degrees of Freedom: {dof_3}")
print(f"\nDecision at α = 0.05:")
if p_3 < 0.05:
    print("Reject H₀  — reorder behaviour is associated with order size category")
else:
    print("Fail to Reject H₀ — reorder behaviour is independent of order size category")

# Verify expected frequencies
print(f"\nMinimum Expected Frequency: {expected_3.min():.2f}")
print(f"All expected frequencies ≥ 5: {(expected_3 >= 5).all()}")

# Contributions
contributions_3 = (np.array(ct3) - expected_3)**2 / expected_3
contrib_df3 = pd.DataFrame(contributions_3,
                            index=ct3.index,
                            columns=ct3.columns)
print("\nContributions per cell:")
print(contrib_df3)
```

**Business Interpretation:**

Cells involving small orders contribute most to the χ² statistic, indicating that their reorder behaviour deviates most from expected under independence.

Large and medium orders show higher reorder rates compared to small orders, suggesting an association between larger basket sizes and more frequent reordering behaviour.

Small orders are more likely to represent occasional or exploratory purchases with relatively more first-time purchases.

From a business perspective, Instacart could target small-order customers with promotions or bundles to encourage increased basket size and higher reorder frequency
