 ### Step 7: Check for Missing Values
We check each dataset for missing values using `.isna().sum()`.  
- This counts the number of `NaN` entries per column.  
- Sorting in descending order highlights columns with the most missing values first.  

**Observations based on the outputs:**
- **Orders:** Only `days_since_prior_order` has missing values — **206,209 NaN entries**
  out of 3,421,083 rows (approximately 6% of the dataset). This is expected because
  a customer's very first order has no previous order to measure from.
- **Products:** No missing values detected across all 4 columns.
- **Order_Products:** No missing values detected across all 4 columns.

```python
# Step 7: Check Missing Values
print("=== Missing Values in Orders Dataset ===\n")
orders_missing = orders.isna().sum().sort_values(ascending=False)
display(orders_missing)

print("\n=== Missing Values in Products Dataset ===\n")
products_missing = products.isna().sum().sort_values(ascending=False)
display(products_missing)

print("\n=== Missing Values in Order_Products Dataset ===\n")
order_products_missing = order_products.isna().sum().sort_values(ascending=False)
display(order_products_missing)
```
